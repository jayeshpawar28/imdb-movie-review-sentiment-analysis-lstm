####Text Preprocessing :

```python
## Remove HTML Tags
## Remove Non-Alpha
## Lower Casing
```

```python
import re
```

```python
re.sub("<.*?>","",df['review'][99])
```

```python
## Removing non-alha chars
re.sub("[^a-zA-Z ]","",df['review'][99])
```

```python
def preprocessing(review):
  ## Lower casing
  review = review.lower()

  ## Remove html tags
  review = re.sub("<.*?>","",review)

  ## Removing non-alha chars
  review = re.sub("[^a-zA-Z ]","",review)

  return review
```

```python
df['review'] = df['review'].apply(preprocessing)
```

```python
df['review'][999]
```

####Tokenization :

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
## Tokenization : Tokenize & Vocab
tokenizer.fit_on_texts(df['review'])
```

```python
tokenizer.word_index
```

```python
len(tokenizer.word_index)
```

```python
## Convert text into input ids
input_ids = tokenizer.texts_to_sequences(df['review'])
```

```python
df['review'][0]
```

```python
input_ids[0]
```

####Embeddings : Converting text into vectors

```python
X = input_ids        ## Features
y = df['sentiment']  ## Target
```

```python
import numpy as np
```

```python
# y = df['sentiment'].replace({'positive':1,'negative':0})
y = np.array([1 if sentiment=="positive" else 0 for sentiment in df['sentiment']])
```

```python
y
```

####Padding :

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
X_padded = pad_sequences(X, maxlen=500)
```

```python
len(X_padded[0])
```

```python
X_padded.shape[1]
```

```python
## Splitting data
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_padded, y,
                                                    test_size=0.2,
                                                    random_state=42
                                                    )
```

####Model Building :

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense,LSTM
```

```python
## input to embedding layer is OHE
input_dim=len(tokenizer.word_index)+1,
#len of vocab + 1 (for padding)

output_dim=100,  ## Dimension of each vector

input_length=len(X_padded[0])  ## Sequence length of each review (2441)
```

```python
model = Sequential()

model.add(Embedding(
    input_dim=len(tokenizer.word_index)+1,
    output_dim=100,
    input_length=X_padded.shape[1]
))

model.add(LSTM(128))

model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
)
```

```python
model.fit(
    X_train,
    y_train,
    epochs=1,
    batch_size=64  ## power of 2
)
```
